```python
## This code compiles a jailbreak from AutoDan24 into compatible csv file for evaluation
source_file = "defense_testing_results/llama2_jbb.json"
output_file = "responses/llama2_jbb.csv"

import json
import pandas as pd

with open(source_file, "r") as f:
    dtr = json.load(f)

columns = [
    "category",
    "source",
    "forbidden_prompt",
    "jailbroken_prompt",
    "jailbreak",
    "raw_response",
    "model",
    "response",
]

rows = []
for record_id in map(str, range(100)):
    record = dtr[record_id]
    rows.append(
        {
            "category": "none",
            "source": "jailbreakbench",
            "forbidden_prompt": record["goal"],
            "jailbroken_prompt": record["user_text_prompt"],
            "jailbreak": "AutoDAN24",
            "raw_response": record["output"],
            "model": "llama2",
            "response": record["output"],
        }
    )

df = pd.DataFrame(rows, columns=columns)
df.to_csv(output_file, index=False)
```

```python
## Setting up data for averine
model_name = "vicuna-7b-v1.5"
dataset = "strong_reject100"
attack = "pair"
results_path = "results/vicuna-7b-v1.5_strong_reject100_pair_finetuned/20261005_160243/responses_and_evaluations.csv"
import json
import pandas as pd
output_file_name = "_".join(results_path.split("/")[1].split("_")[0:-1])
print(output_file_name)
df = pd.read_csv(results_path)
data = []
for row in df.itertuples():
    data.append(
        {
            "index": row.Index,
            "forbidden_prompt": row.forbidden_prompt,
            "jailbroken_prompt": row.jailbroken_prompt,
            "jailbreak": row.jailbreak,
            "model": row.model,
            "response": row.response,
        }
    )
output = {
    "model_name" : model_name,
    "dataset" : dataset,
    "attack" : attack,
    "data" : data
}

json.dump(output, open(f"data/averine/{output_file_name}.json", "w"), indent=4)
```
