```python
## This code compiles a jailbreak from AutoDan24 into compatible csv file for evaluation
source_file = "defense_testing_results/vicuna_jbb.json"
output_file = "responses/vicuna_jbb.csv"

import json
import pandas as pd

with open(source_file, "r") as f:
    dtr = json.load(f)

columns = [
    "category",
    "source",
    "forbidden_prompt",
    "jailbroken_prompt",
    "jailbreak",
    "raw_response",
    "model",
    "response",
]

rows = []
for record_id in map(str, range(100)):
    record = dtr[record_id]
    rows.append(
        {
            "category": "none",
            "source": "jailbreakbench",
            "forbidden_prompt": record["goal"],
            "jailbroken_prompt": record["user_text_prompt"],
            "jailbreak": "AutoDAN24",
            "raw_response": record["output"],
            "model": "vicuna",
            "response": record["output"],
        }
    )

df = pd.DataFrame(rows, columns=columns)
df.to_csv(output_file, index=False)
```
